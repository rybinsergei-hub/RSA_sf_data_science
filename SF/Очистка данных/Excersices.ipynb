Задание 2.4

Проверим, влияет ли уровень экологической обстановки в районе на цену квартиры. Постройте коробчатую диаграмму цен на квартиры (price_doc)\
в зависимости от уровня экологической обстановки в районе (ecology).\

Какой уровень ценится на рынке меньше всего?

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig = plt.figure(figsize=(10, 5))
boxplot = sns.boxplot(
    data=sber_data,
    x='ecology',  # категориальная переменная на оси X
    y='price_doc',  # числовая переменная на оси Y
    width=0.9
)
boxplot.set_title('Влияние экологической обстановки на цену')
boxplot.set_xlabel('Экология')
boxplot.set_ylabel('Цена')
boxplot.grid()

plt.show()
```

Задание 2.5

Постройте диаграмму рассеяния, которая покажет, как цена на квартиру (price_doc) связана с расстоянием до центра Москвы (kremlin_km). Выберите все верные утверждения.

```python
fig = plt.figure(figsize=(10, 7))
scatter = sns.scatterplot(
    data=sber_data,
    x='kremlin_km',
    y='price_doc',
    alpha=0.6  # Прозрачность для лучшего отображения множества точек
)

scatter.set_title('Зависимость цены квартиры от расстояния до центра Москвы')
scatter.set_xlabel('Расстояние до Кремля (км)')
scatter.set_ylabel('Цена квартиры')
scatter.grid()

plt.show()
```
